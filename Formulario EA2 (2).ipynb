## Matrices y aplicaciones

```python
#Definición de matrices en Python
A = np.array([[elemento, elemento, elemento, elemento],
             [elemento, elemento, elemento, elemento],
             [elemento, elemento, elemento, elemento]])
B = np.array([[elemento, elemento, elemento, elemento],
             [elemento, elemento, elemento, elemento],
             [elemento, elemento, elemento, elemento]])
print(f'La matriz A es:\n\n{A}')
print(f'La matriz B es:\n\n{B}')
```

```python
#Suma de matrices
S = A + B
print(f'La matriz S es:\n\n{S}')
```

```python
#resta de matrices
R = B - A
print(f'La matriz R es:\n\n{R}')
```

```python
#Mutiplicación de matrices por un escalar
E = 0.8 * C
print(f'La producción de Ecuador es:\n\n{E}')
```

```python
#Multiplicación de matrices
C = np.dot(A, B)
print(f'La matriz C es:\n\n{C}.')
```

```python
#Transpuesta de una matriz
T = B.T
print(f'La matriz transpuesta de T es:\n\n{T}')
```

```python
#Cración de matrices mediante función sobre los índices

D = np.fromfunction(lambda i, j: función sobre los índices, (orden de la matriz))

print(f'La matriz D es:\n\n{D}\n\n')
```

```python
#Inversa de una matriz

F = np.array([[elemento, elemento],
              [elemento, elemento]])

F_inv = np.linalg.inv(F)
print(f'La inversa de la matriz F es:\n\n{F_inv}\n\n')
```
