# Problema 1

Un equipo de desarrollo está evaluando la eficiencia de un nuevo algoritmo implementado en *Python*. Han medido el tiempo de ejecución del algoritmo en función de la cantidad de elementos de entrada, obteniendo los resultados que se muestran en la siguiente tabla:

<p><center>

| Cantidad de elementos (unidades) | Tiempo de ejecución (ms) |
|----------------------------------|--------------------------|
|$$ 100 $$                         | $$2 $$                   |
|$$ 200 $$                         | $$4 $$                   |
|$$ 500 $$                         | $$10 $$                  |
|$$ 1.000 $$                        | $$20 $$                  |
|$$ 2.000 $$                        | $$40 $$                  |

</center></p>

1. Defina variable dependiente e independiente, indicando unidad de medida.
2. Encuentra la función que relaciona el tamaño de la entrada con el tiempo de ejecución del algoritmo, para esto utiliza la función *polyfit* de *Numpy*.
3. Calcula el tiempo de ejecución para una entrada de $1.500$ elementos.
4. Utiliza la función *fsolve* de *Scipy* para determinar el tamaño de entrada que haría que el tiempo de ejecución sea de $50$ milisegundos.


### 1) Definir dependiente e independiente

Variable dependiente = Tiempo en milisegundos (ms)

Variable independiente = Unidades de elementos

### 2) Encontrar función 

```python
x = np.array([100, 200, 500, 1000, 2000])
y = np.array([2,4,10,20,40])
pendiente, intercepto = np.polyfit(x,y,1)
print(f"el valor de la pendiente es {pendiente:.2f} y el valor del intercepto es {intercepto:.2f}.")
```

$t(n)$ $=$ $0.02n$+$0$

### 3)Calcular 1500 elementos

```python
def t(n):
    return 0.02* n

print(f"La ejecución de entrada para 1500 elementos es de {t(1500)} ms.")
```

### 4) Tamaño de entrada para 50 ms

```python
def f(n):
    return 0.02*n-50

n = np.linspace(0,2000,1)
sol = fsolve(f,n)
print(f"Si el tiempo de ejecución es de 50 ms, la cantidad de elementos de entrada es de {sol} unidades.")
```

3. Interprete la pendiente en el contexto del problema.
4. ¿Cuánto tiempo toma una transferencia de $73,2$ GB?
5. Utiliza la función *fsolve* de *Scipy* para determinar la cantidad de datos que se puedan transferir en $123,5$ minutos.

1. Encuentra las funciones que modelan el tiempo de carga $f(x)$ y el tiempo de descarga $g(x)$ en función del tamaño del archivo $x$. Utiliza la función *polyfit* de *Numpy* (debes encontrar dos funciones).

3. Calcula el tiempo de carga y descarga para un archivo de $750$ MB.

### 1) Modelar funciones f(x) y g(x)

```python
archivo = np.array([50, 100, 250, 500, 1000])
carga = np.array([10, 20, 50, 100, 200])
descarga = np.array([8, 16, 40, 80, 160])
pendiente1, intercepto1 = np.polyfit(archivo, carga,1)
pendiente2, intercepto2 = np.polyfit(archivo, descarga,1)
print(f"f(x):el valor de la pendiente es {pendiente1:.2f} y el valor del intercepto es {intercepto1:.2f}.")
print(f"g(x):el valor de la pendiente es {pendiente2:.2f} y el valor del intercepto es {intercepto2:.2f}.")
```

$f(x)$=$0.20x$


$g(x)$=$0.16x$

### 2) Variable independiente y dependiente

$f(x)$---> V.i= Tamaño de los archivos en MB
v.d = Tiempo de carga en segundos



$g(x)$----> v.i =Tamaño de los archivos en MB
v.d= Tiempo de descarga en segundos

### 3)Tiempo de carga y de descarga en un archivo de 750 MB

```python
def f(x):
    return 0.20*x

def g(x):
    return 0.16*x


print(f"Si el archivo tiene un tamaño de 750 MB, el tiempo de carga es de {f(750):.2f} s y el de descarga es de {g(750):.2f} s.")
```

### 4) Determinar el tamaño de un archivo que demora 163 s de carga

```python
def C(x):
    return 0.20*x-163

x = np.linspace(0,1000,1)
sol = fsolve(C,x)
print(f"Si el tiempo de carga es de 163 s, el tamaño del archivo es de {sol} MB.")
```

### 5) Tamaño de un archivo que demora 195 en descargar

```python
def D(x):
    return 0.16*x-195

x = np.linspace(0,1000,1)
sol = fsolve(D,x)
print(f"Si el tiempo de descarga es de 195 s, el tamaño del archivo es de {sol} MB.Entonces el trabajor estaba equivocado.")
```

### 6) Graficar

```python
def f(x):
    return 0.20*x

def g(x):
    return 0.16*x

x = np.arange(0, 1000.1, 0.1)
plt.plot(x, f(x), label = 'Carga')
plt.plot(x, g(x), label = 'Descarga')
plt.title('Relación entre el tamaño de un archivo en MB y el tiempo en segundos')
plt.xlabel('Tiempo (s)')
plt.ylabel('Tamaño del archivo (MB)')
plt.legend()
plt.show()
```
