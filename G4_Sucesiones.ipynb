# Problema 1

Considere la sucesión $a_{n} = 3n^{2} + 7$. Por medio de un código en *Python* despliega en pantalla:

1. Los primeros $5$ términos.
2. Los $5$ términos que vienen inmediatamente después del decimoquinto término.

### 1)

```python
a = []
for i in range(20):
  n = i + 1
  a.append(3*n**2+7)
  if n == 1:
    print('1) Los primeros 5 términos de la sucesión son:\n')
  elif n == 16:
    print(' ')
    print('2) Los términos que vienen inmediatamente después del 15 son:\n')

  if n < 6 + 1 or n > 15:
    print(f'a_{n} = {a[i]}')
```

```python
final=40000
termino=0
n=0
dif=final-termino
while dif>0:
    n +=1
    termino=5*n**3
    dif=final-termino

if dif==0:
    print(f'El {final} es un término de la sucesión y ocupa el lugar {n}')

else:
    print(f'El {final} no pertenece a la sucesión')
```
