```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(t):
    return 2*t**2+5*t-3
```

```python
t = np.arange(0,20,0.1)
plt.plot(t,f(t),label = "f(t)")
plt.title("Relación de la velocidad en función del tiempo")
plt.xlabel("Tiempo transcurrido en segundos")
plt.ylabel("Velocidad en m/s")
plt.legend()
plt.grid(True)
plt.show()
```
