Una empresa de redes vende dos tipos de conectores para confeccionar cables de red. El conector R45cat5 se vende a $700 la unidad y el conector R45cat6 se vende a $1.200 la unidad. Para asegurar una rentabilidad mínima en la empresa, el número total de conectores vendidos diariamente debe ser 60 y para lograr la venta objetivo diaria se debe asegurar obtener un ingreso de $60.000 por la venta de conectores. La situación anterior, se puede modelar a través del siguiente sistema de ecuaciones:

<br>

\begin{split}
700x + 1200y &= 60000 \\
x + y &= 60
\end{split}

¿Cuál es la cantidad de conectores R45cat5 y R45cat6 que se deben vender para asegurar la rentabilidad mínima y lograr la venta objetivo diaria?

```python
import numpy as np
```

### 1) Represente matricialmente el sistema de ecuaciones.

<br>

\begin{split}
700x + 1200y &= 60000 \\
x + y &= 60
\end{split}

<br>

\begin{split}
\begin{pmatrix}
 700  &  1200\\
 1  &  1 \\
\end{pmatrix} \cdot \begin{pmatrix}
 x \\
 y \\
\end{pmatrix} = \begin{pmatrix}
 60000 \\
  60\\
\end{pmatrix}
\end{split}

<br>

\begin{split}
\begin{pmatrix}
 x \\
 y \\
\end{pmatrix} = \begin{pmatrix}
 700  &  1200 \\
 1  &  1\\
\end{pmatrix} ^{-1}\cdot \begin{pmatrix}
 60000 \\
 60 \\
\end{pmatrix}
\end{split}

<br>

### 2) Encuentra los valores de las incógnitas (utilizando los códigos en Python vistos en clases).

```python
A = np.array([[700, 1200],
              [1, 1]] )

B = np.array([[60000],
              [60]] )

A_inv = np.linalg.inv(A)

C = np.dot(A_inv,B)
print(f'La matriz de incógnitas es:\n\n{C}')
```
